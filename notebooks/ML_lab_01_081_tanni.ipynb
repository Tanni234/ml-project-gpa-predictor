```python
import numpy as np
np.random.seed(42)
marks = np.array([55, 72, 48, 90, 66, 81])
print('Marks:', marks)
print('Mean:', marks.mean())
print('Highest:', marks.max())
print('Passed:', marks[marks >= 50])
```

```python
import numpy as np

marks = np.array([45, 62, 78, 35, 90, 55])

min_mark = np.min(marks)
std_dev = np.std(marks)  # Population standard deviation

print("Minimum Mark:", min_mark)
print(f"Standard Deviation: {std_dev:.2f}")
```

```python
import pandas as pd
data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}
df = pd.DataFrame(data)
display(df)
print(df.info())
print(df.select_dtypes(include='number').describe())
```

```python
# 18. Attendance >= 80
high_attendance = df[df['attendance'] >= 80]
print(high_attendance)

# 19. Average study hours for passed students
avg_study_passed = df[df['passed'] == 'Yes']['study_hours'].mean()
print('Avg study hours (passed):', avg_study_passed)

# 20. Sort by attendance descending
sorted_df = df.sort_values('attendance', ascending=False)
print(sorted_df)
```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
    return np.random.randint(40, 101, size=5)

print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```
